# 02. Exploratory Data Analysis (EDA)
In this notebook, we analyze closing cutoff percentiles across academic years, engineering branches, admission CAP rounds, and seat categories.


```python
# Step 1: Load normalized cutoff dataset and compute summary statistics
import pandas as pd
import matplotlib.pyplot as plt

df_cut = pd.read_csv('../dataset/processed/cutoffs_long.csv')
print(f'Total records: {len(df_cut):,}')
print(df_cut['cutoff_percentile'].describe())
```

```python
# Step 2: Branch-wise median cutoffs
print('Branch-wise Median Cutoff Percentiles:')
print(df_cut.groupby('branch_name')['cutoff_percentile'].median().sort_values(ascending=False))
```

```python
# Step 3: Visualizing Cutoff Distribution
from PIL import Image
plt.figure(figsize=(10, 5), dpi=120)
img = Image.open('../report/screenshots/eda_cutoff_distribution.png')
plt.imshow(img)
plt.axis('off')
plt.show()
```

```python
# Step 4: Visualizing Branch-wise Boxplot
plt.figure(figsize=(12, 6), dpi=120)
img2 = Image.open('../report/screenshots/eda_branch_boxplot.png')
plt.imshow(img2)
plt.axis('off')
plt.show()
```
