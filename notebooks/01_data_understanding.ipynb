In this notebook, we explore the official regulatory documents published by the State Common Entrance Test Cell, Maharashtra State for undergraduate engineering CAP admissions.


```python
# Step 1: Inspect official document audit manifest
import pandas as pd

df_manifest = pd.read_csv('../docs/source_manifest.csv')
print(f'Total regulatory source documents: {len(df_manifest)}')
print(df_manifest[['academic_year', 'document_type', 'cap_round', 'file_name', 'status', 'file_size_bytes']])
```

```python
# Step 2: Inspect raw text extracted per PDF page
df_pages = pd.read_csv('../dataset/interim/pdf_pages.csv')
print(f'Total interim extracted pages: {len(df_pages):,}')
print('\nPage breakdown by document type and academic year:')
print(df_pages.groupby(['academic_year', 'document_type'])['page_no'].count())
```

```python
# Step 3: Inspect sample raw text lines to observe DTE 10-digit course pattern
sample_text = df_pages['text'].iloc[0]
print('Sample extracted text snippet (Page 1):')
print('\n'.join(sample_text.split('\n')[:15]))
```
